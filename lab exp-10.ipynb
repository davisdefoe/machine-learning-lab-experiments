```

import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Step 1: Generate sample data
np.random.seed(42)

X1 = np.random.normal(5, 1, (100, 2))
X2 = np.random.normal(10, 1, (100, 2))

X = np.vstack((X1, X2))

# Step 2: Create Gaussian Mixture Model
gmm = GaussianMixture(
    n_components=2,
    random_state=42,
    max_iter=100
)

# Step 3: Fit the model using EM algorithm
gmm.fit(X)

# Step 4: Predict clusters
labels = gmm.predict(X)

# Step 5: Display results
print("Expectation-Maximization Algorithm")

print("\nNumber of Iterations:", gmm.n_iter_)
print("Converged:", gmm.converged_)

print("\nCluster Means:")
print(gmm.means_)

print("\nCluster Probabilities:")
print(gmm.weights_)

print("\nPredicted Cluster Labels:")
print(labels[:20])

# Step 6: Visualize clusters
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis')

plt.title("EM Algorithm - Gaussian Mixture Model")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
