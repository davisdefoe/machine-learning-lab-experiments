```

import csv
import os

# STEP 1: Create CSV file automatically
filename = "training_data.csv"

training_data = [
    ["Sky", "AirTemp", "Humidity", "Wind",
     "Water", "Forecast", "EnjoySport"],

    ["Sunny", "Warm", "Normal", "Strong",
     "Warm", "Same", "Yes"],

    ["Sunny", "Warm", "High", "Strong",
     "Warm", "Same", "Yes"],

    ["Rainy", "Cold", "High", "Strong",
     "Warm", "Change", "No"],

    ["Sunny", "Warm", "High", "Strong",
     "Cool", "Change", "Yes"]
]

with open(filename, "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(training_data)

print("CSV file created successfully!")

# STEP 2: Read CSV file
with open(filename, "r") as file:
    data = list(csv.reader(file))

attributes = data[0][:-1]
examples = data[1:]

n = len(attributes)

# STEP 3: Initialize S and G
S = ['0'] * n
G = [['?'] * n]

# Check whether hypothesis h covers example x
def covers(h, x):
    return all(h[i] == '?' or h[i] == x[i]
               for i in range(n))

# Check whether hypothesis h1 is more general than h2
def more_general(h1, h2):
    return all(h1[i] == '?' or h1[i] == h2[i]
               for i in range(n))

# STEP 4: Candidate-Elimination Algorithm
for example in examples:

    x = example[:-1]
    target = example[-1]

    print("\nTraining Example:", x, target)

    if target == "Yes":

        # Generalize S
        if S == ['0'] * n:
            S = x.copy()
        else:
            for i in range(n):
                if S[i] != x[i]:
                    S[i] = '?'

        # Remove hypotheses from G that do not cover positive example
        G = [g for g in G if covers(g, x)]

    else:

        # Specialize G for negative example
        new_G = []

        for g in G:

            if covers(g, x):

                for i in range(n):

                    if g[i] == '?' and S[i] != '?' and S[i] != '0':
                        if S[i] != x[i]:

                            h = g.copy()
                            h[i] = S[i]

                            if covers(h, x):
                                continue

                            new_G.append(h)

            else:
                new_G.append(g)

        G = new_G

    # Remove hypotheses more specific than another hypothesis in G
    G = [
        g for g in G
        if not any(
            g != other and more_general(other, g)
            for other in G
        )
    ]

    print("Specific Boundary (S):", S)
    print("General Boundary (G):", G)

# STEP 5: Display final results
print("\n========== FINAL RESULT ==========")

print("\nSpecific Boundary (S):")
print(S)

print("\nGeneral Boundary (G):")
for g in G:
    print(g)

print("\nCandidate-Elimination completed successfully!")
```
